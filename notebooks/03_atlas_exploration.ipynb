## Avaliação da granularidade

O Atlas disponibiliza dados em diferentes níveis geográficos.

Após a análise, optou-se por utilizar apenas a granularidade de Unidade da
Federação (UF), por apresentar compatibilidade com a modelagem da camada
Silver do projeto.

```python
df_atlas['AGREGACAO'].value_counts()
```

```python
df_atlas_uf = (
    df_atlas
    .query("AGREGACAO == 'Unidade da Federação'")
    .copy()
)

df_atlas_uf.shape
```

## Seleção dos indicadores

## Dataset final

Após a exploração da estrutura da base, foram aplicadas as regras de transformação definidas em src/atlas/transform.py. Essa etapa padroniza a estrutura dos dados, seleciona os indicadores utilizados no projeto, converte os códigos IBGE em siglas de UF e produz a versão tratada (atlas_uf.parquet) utilizada pela pipeline de ingestão.

Após o tratamento, a base ficou composta por 54 registros (27 UFs × 2 anos),
contendo apenas as informações necessárias para enriquecimento da camada
Silver.


```python
from src.atlas.transform import transform_atlas_uf

df_atlas = pd.read_excel("../data/external/adh_radar_base_2012_2024.xlsx")
atlas_uf = transform_atlas_uf(df_atlas)

atlas_uf
```

```python
atlas_uf = atlas_uf[
    atlas_uf["ano"].isin([2023, 2024])
].copy()

atlas_uf.info()
```

## Exportação

A versão tratada é exportada em formato Parquet e utilizada pela pipeline de
ingestão do Atlas.

```python
atlas_uf.to_parquet(
    PROJECT_ROOT / "data" / "external" / "atlas_uf.parquet",
    index=False,
)
```
